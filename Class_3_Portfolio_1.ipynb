## Loops
Loops in Python are used to execute a block of code repeatedly. We will look at two different types of loops; the *for loop*, and the *while loop*.  
**The for loop** iterates over sequences, so it keeps executing until it has reached the end of the loop.  
**The while loop** keeps repeating while a condition is true, and first ends when the condition becomes false. 

```python
# For loop - Basic syntax
# for element in sequence:
    # code to execute for each element

# Example of for loop
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
```

```python
# Exercise - Modify the code to print the length of each fruit instead of the fruit itself.

fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(len(fruit))
```

```python
# While loop - Basic syntax
# while condition:
    # code to execute while condition is True
    
# Example of while loop
x = 5
while x > 0:
    print(x)
    x -= 1
```

```python
# Excercise - Modify the code to count up from 1 to 5 instead of counting down from 5 to 1.

x = 1
while x <= 5:
    print(x)
    x += 1
```

You’re making a simple password checker. How would you check if the password is correct?

Write code that asks for a password and checks if it follows the requirements:
- It needs to be between 8 and 32 characters long 

Make sure you print some meaningful message the user can understand. Try it with both the correct password and wrong passwords.

Hint: use the input function to allow for user inputs e.g. **input("Enter the password: ")**

```python
# the input() isn't supported by my jupyter notebook :(

def password_checker(password):
    
    if len(str(password)) >= 8 and len(str(password)) <= 32:
         print("Your password is correct!")
        
    elif len(str(password)) < 8:
        print("Your password is too short - It needs to be at least 8 characters long")
        
    else:
        print("Your password is too long - It can be a maximum of 32 characters long")

# checking passwords: 

password_checker("aarhus")

password_checker("culturaldatascience")

password_checker("culturaldatascienceculturaldatascienceculturaldatascience")
```

Now lets add some extra requirements:
- The password needs to have at least 3 digits
- And there needs to be an uppercase letter

```python
def password_checker(password):
    
    if len(password) >= 8 and len(password) <= 32:
        if any(letter.isupper() for letter in password) == True:
            if sum(digit.isdigit() for digit in password) >= 3:
                print("Your password is correct!")

            else: 
                print("Your password needs to have at least three digits")      
        else: 
            print("Your password needs to have at least one uppercase letter")   
    elif len(password) < 8:
        print("Your password is too short - It needs to be at least 8 characters long") 
    else:
        print("Your password is too long - It can be a maximum of 32 characters long")

# checking passwords: 

password_checker("sarahsarah")

password_checker("SarahSarah")

password_checker("SarahSarah1010")
```

And the last bits, now we also need to ensure that the password contains:
- 2 special characters

Make sure the user understands why a wrong password is rejected. And ensure that only a password that holds all requirements are passed. 

```python
special_characters = ['$', '#', '@', '!', '*', '€', '?', '+', '-', '<', '>', '^']

def password_checker(password):
    
    if len(password) >= 8 and len(password) <= 32:
        if any(letter.isupper() for letter in password) == True:
            if sum(digit.isdigit() for digit in password) >= 3:
                if any(characters in special_characters for characters in password) == True:
                    print("Your password is correct!")

                else: 
                    print("Your password needs to have at least one special character")
            else: 
                print("Your password needs to have at least three digits")               
        else: 
            print("Your password needs to have at least one uppercase letter")
    elif len(password) < 8:
        print("Your password is too short - It needs to be at least 8 characters long")
    else:
        print("Your password is too long - It can be a maximum of 32 characters long")

# checking passwords: 

password_checker("sarahsarah")

password_checker("SarahSarah")

password_checker("SarahSarah1010")

password_checker("SarahSarah1010!")
```

## Temperature converter

Lets now make a function that can convert celcius to fahrenheit. 
Make an IPO with your group and discuss what the input, the process and the output should be. 

The function should print an informational message and return something usable. 

Hint: F = (celsius * 9/5) +32

```python
# celsius to fahrenheit converter

def celsius_to_fahrenheit_converter(celsius):
    fahrenheit = (celsius * 9/5) + 32
    print(f'{celsius} degrees celcius is {fahrenheit} degrees fahrenheit.')

celsius_to_fahrenheit_converter(32)
```

Now make a function that does the opposite.   
Hint: C = ((fahrenheit-32)/(9/5))  

```python
# fahrenheit to celsius converter

def fahrenheit_to_celsius_converter(fahrenheit):
    celsius = ((fahrenheit - 32)/(9/5))
    print(f'{fahrenheit} degrees fahrenheit is {celsius} degrees celsius.')

fahrenheit_to_celsius_converter(105)
```

Well done! 
Now wouldn't it be neat if we could just wrap that into a function that we could call temperature_converter with two parameters "temperature" and "unit" so that we could convert both ways. 



```python
# Temperature converter here

def temperature_converter(temperature, unit):
    
    if unit == "celsius":
        fahrenheit = (temperature * 9/5) + 32
        print(f'{temperature} degrees celsius is {fahrenheit} degrees fahrenheit.')

    elif unit == "fahrenheit":
        celsius = ((temperature - 32)/(9/5))
        print(f'{temperature} degrees fahrenheit is {celsius} degrees celsius.')

temperature_converter(32, "celsius")

temperature_converter(105, "fahrenheit")
```

## Grade calculator

The scenario: You’re helping a teacher who wants to calculate student grades quickly. The teacher has been doing these calculations by hand, which takes a long time and sometimes leads to mistakes.

The challenge: 
- Calculate the average of test scores
- Convert percentage grades to letter grades (it's just easier than the danish grading metric ngl)
- Calculate final grades with different weights for different assignments

Example data:  
Tests count for 70%, homework counts for 30%  

Test scores: 85, 93, 78, 96, 88
Homework average: 90

Now we start with one thing first. Each functions should do _one_ thing effectively. 


```python
# Make a function that calculate the average of test scores. 

def calculate_average(*args):
    average = (sum(args))/len(args)
    print(f'The average of the score is {average}%')

# testing it out

calculate_average(85, 93, 78, 96, 88)
```

Well done, now we move on to the next sub-task

Hint: Grade to letter grade converter  
90–100 → A  
80–89  → B  
70–79  → C  
60–69  → D  
0–59   → F  


```python
# Make a function that converts the test scores to letter grades

def percentage_to_letter_grade(test_score):

    if test_score >= 90:
        grade = "A"

    elif test_score >= 80 and test_score <= 89:
        grade = "B"

    elif test_score >= 70 and test_score <= 79:
        grade = "C"

    elif test_score >= 60 and test_score <= 69:
        grade = "D"

    else: 
        grade = "F"

    print(f'With a score of {test_score}%, the grade is {grade}')
    
# Test with different percentages

percentage_to_letter_grade(58)

percentage_to_letter_grade(60)

percentage_to_letter_grade(74)

percentage_to_letter_grade(82)

percentage_to_letter_grade(91)
```

Last task, We need to calculate the weighted grade


```python
#Tests count for 70%, homework counts for 30%

def calculate_weighted_grade(test_average, homework_average):
    weighted_average = (test_average * 0.70) + (homework_average * 0.30)
    print(f'The weighted grade is {weighted_average}%')  

calculate_weighted_grade(88, 90)

percentage_to_letter_grade(88.6)
```

Now we want to give the professor one function that uses all these three and wraps them in a neat little function with everything put together. 

```python
def calculate_student_grade(homework_average, *args):
    test_average = (sum(args))/len(args)

    weighted_average = (test_average * 0.70) + (homework_average * 0.30)

    if weighted_average >= 90:
        grade = "A"

    elif weighted_average >= 80 and weighted_average <= 89:
        grade = "B"

    elif weighted_average >= 70 and weighted_average <= 79:
        grade = "C"

    elif weighted_average >= 60 and weighted_average <= 69:
        grade = "D"

    else: 
        grade = "F"

    print(f'Final Letter Grade: {grade}\n\nWeighted Average: {weighted_average}%\n\nTest Average: {test_average}%\n\nTest Scores: {args}\n\nHomework Average: {homework_average}%') 

calculate_student_grade(90, 85, 93, 78, 96, 88)
```
