**While-loops:**

```python
sweet_enough = 3
num_sugar_spoons = 0
while num_sugar_spoons < sweet_enough:
    num_sugar_spoons = num_sugar_spoons + 1
    print('I have added', num_sugar_spoons, 'spoons of sugar')
```
