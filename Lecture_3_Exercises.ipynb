```python
# define book class (structure for containing book information).
class Book:
    def __init__(self, title, author, year, genre, country):
        self.title = title
        self.author = author
        self.year = year
        self.genre = genre
        self.country = country

    def describe_book(self):
        print(f"The book {self.title} is written by {self.author} from {self.country} in {self.year}, and the genre is {self.genre}")
```

```python
# 1. Create a dictionary with 20 objects of the Book class. 
        # hint: 
        # books_db = {
   # "1984": Book("1984", "George Orwell", 1949, "Dystopian", "UK"),
    #"To Kill a Mockingbird": Book("To Kill a Mockingbird", "Harper Lee", 1960, "Fiction", "USA")...}

book_db = {
    "1984": Book("1984", "George Orwell", 1949, "Dystopian", "UK"),
    "The Hunger Games": Book("The Hunger Games", "Suzanne Collins", 2008, "Young Adult", "USA"),
    "To Kill a Mockingbird": Book("To Kill a Mockingbird", "Harper Lee", 1960, "Fiction", "USA"),
    "Pride and Prejudice": Book("Pride and Prejudice", "Jane Austen", 1813, "Romance", "UK"),
    "Harry Potter": Book("Harry Potter", "J.K. Rowling", 1997, "Fantasy", "UK"),
    "The Hobbit": Book("The Hobbit", "J.R.R. Tolkien", 1937, "Fantasy", "UK"),
    "The Great Gatsby": Book("The Great Gatsby", "F. Scott Fitzgerald", 1925, "Fiction", "USA"),
    "The Catcher in the Rye": Book("The Catcher in the Rye", "J.D. Salinger", 1951, "Fiction", "USA"),
    "Brave New World": Book("Brave New World", "Aldous Huxley", 1932, "Dystopian", "UK"),
    "The Alchemist": Book("The Alchemist", "Paulo Coelho", 1988, "Adventure", "Brazil"),
    "The Little Prince": Book("The Little Prince", "Antoine de Saint-Exupéry", 1943, "Fable", "France"),
    "Don Quixote": Book("Don Quixote", "Miguel de Cervantes", 1605, "Adventure", "Spain"),
    "The Book Thief": Book("The Book Thief", "Markus Zusak", 2005, "Historical Fiction", "Australia"),
    "Life of Pi": Book("Life of Pi", "Yann Martel", 2001, "Adventure", "Canada"),
    "The Kite Runner": Book("The Kite Runner", "Khaled Hosseini", 2003, "Historical Fiction", "Afghanistan"),
    "The Fault in Our Stars": Book("The Fault in Our Stars", "John Green", 2012, "Young Adult", "USA"),
    "The Da Vinci Code": Book("The Da Vinci Code", "Dan Brown", 2003, "Mystery", "USA"),
    "The Road": Book("The Road", "Cormac McCarthy", 2006, "Post-apocalyptic", "USA"),
    "Little Women": Book("Little Women", "Louisa May Alcott", 1868, "Fiction", "USA"),
    "The Picture of Dorian Gray": Book("The Picture of Dorian Gray", "Oscar Wilde", 1890, "Gothic", "Ireland")
}

book_db["The Hunger Games"].describe_book()

book_db["Little Women"].describe_book()
```

```python
# 2. Define a function that find_books_by_genre(genre) that allows for searching the dictionary by genre.

def find_books_by_genre(genre):
    for book in book_db.values():
        if book.genre == genre:
            book.describe_book()

find_books_by_genre("Fantasy")

find_books_by_genre("Dystopian")
```

```python
# 3. Define a function that count_books_before_year(year) that allows for counting the number of books 
# written before a specific year. 

def count_books_before_year(year):
    count = 0

    for book in book_db.values():
        if book.year < year:
            count += 1

    return count

print(count_books_before_year(1943))

print(count_books_before_year(2005))
```
