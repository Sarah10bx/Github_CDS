## ***Container Variables***

```python
# making lists:

shopping = ["chocolate", "coffee", "milk", "apples"]
scores = [23, 63, 79, 21]
mixed = ["dogs", 7, 2, 3]
combined = shopping, scores, mixed

# lists are mutable:

shopping[0] = "caramel"

print(shopping)
print(scores)
print(mixed)
print(combined)
```

```python
# trying out some indexing

print(shopping[0])
print(shopping[1])
print(shopping[0:2])
```

```python
# now we're going to make some tuples: 

shopping_tuple = ("chocolate", "coffee", "milk", "apples")

# tuples are not mutable
```

```python
# there's also sets, which is an unordered set of unique items: 

shopping = ["chocolate", "coffee", "milk", "apples", "apples", "coffee"]

unique_shopping_set = set(shopping)

print(unique_shopping_set)

# check whether a set contains an element:

check_apples = "apples" in unique_shopping_set
print(check_apples)

check_bobs = "bobs" in unique_shopping_set
print(check_bobs)
```
