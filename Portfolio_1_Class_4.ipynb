## Data organisation using a dataset about flowers

```python
flower = datasets.load_iris()

# convert to DataFrame
df = pd.DataFrame(flower.data, columns=flower.feature_names)

df.head()
```


**Lets take a look at the data frame**   

```python
# Commands in pandas that can give you a quick overview of the data:

df.head()      # first 5 rows, use paranthesis you can decide rows
df.tail()      # last 5 rows
df.info()      # summary of columns and types
df.describe()  # quick statistics (for numbers)
```

**Selecting columns and rows**

Try to run the cell below and figure out which output is linked to the code 

```python
# If you want to select a specific column, select it using name:
print(df['sepal length (cm)'].head())

# If you would like to print one row, use the index of the row:
print(df.iloc[0])

# if you want to select a few rows of only a few columns:
print(df.iloc[0:3 , 0:2])  # first three rows, first two columns

# And if you want to select specific data, 
# you can specify a single row and column:
print(df.iloc[2,0])  # second row, first column

# Or use the column name:
print(df.loc[2, "sepal length (cm)"]  )
```

**Subsetting data**   
Subsetting is the process of retrieving just the parts of large files which are of interest for a specific purpose.   
This will come in handy for your projects when you have to work with potentially large data files 

```python
# Let's try to select some data using conditionals

# Select all rows where sepal length is >= to 5 cm:

lengt_above_five = df[df["sepal length (cm)"] >= 5]   
lengt_above_five.head()
```

```python
# Rows where sepal length is >= to 5 cm and the sepal width < 3,5 cm:

length_and_width = df[(df["sepal length (cm)"] >= 5
                      ) & (df["sepal width (cm)"] < 3.5)]
length_and_width.head()
```

```python
# Excercise - Find longest petal length and median petal length

#use .max() function on petal length
longest_petal_length = df['petal length (cm)'].max()
# print message
print(f"The longest petal length is {longest_petal_length} cm")

#use .median() function on petal length
median_petal_length = df['petal length (cm)'].median()
# print message
print(f"The median petal length is {median_petal_length} cm") 

# subset flowers between the median and one cm shorter than max
subsest_median_and_max = df[(df["petal length (cm)"] 
                             >= median_petal_length) & (
    df["petal width (cm)"] < (longest_petal_length-1))]

# print first 5 rows:
subsest_median_and_max.head()
```

**Joining**
Sometimes we have multiple dataframes we woudl like to add together. Maybe you have been subsetting parts of an old dataframe to substract important information and would now like join them so you can begin your analysis. 

```python
# Joining a bit of the iris data with a new dataframe 
# We will make up some data here:

# selecting the first 10 rows of the iris data
first_10 = df.iloc[0:10, :] 

new_data = {"color": ["red", "blue", "green", "yellow", 
                      "purple", "red", "blue", "green", 
                      "yellow", "purple"],
            "height": [80, 80, 70, 100, 90, 80, 80, 70, 
                       100, 90]}

# new_df is a dictionary, convert it to a dataframe

new_df = pd.DataFrame(new_data)

#Now we join the two dataframes

# There are 4 different types of how: outer, inner, left, right. 
joined_left = first_10.join(new_df, how='left')

joined_left
```

*Different types of how to join two data frames*  
This is important if your dataframes do not have the same amount of rows

left → all rows from the left DataFrame - and the ones matching from the right.

right → all rows from the right DataFrame - and the ones matching from the right.

inner → only rows with matching index values (rows) in both.

outer → all rows from both, fill missing with NaN.


```python
# Excercise - Which types of join will work? 
# Try them out and see what happens

# joined right: 

joined_right = first_10.join(new_df, how='right') 

joined_right

# joined outer: 

joined_outer = first_10.join(new_df, how='outer') 

joined_outer

# joined inner: 

joined_inner = first_10.join(new_df, how='inner') 

joined_inner
```

**All of them has the excact same out. Perhaps, because they each have the same amount of rows.**

```python
# Excercise 2 - Add a row to one of the dataframes
# Then see what happens when you join them again

new_data_2 = {"color": ["red", "blue", "green", "yellow", 
                        "purple", "red", "blue", "green", 
                        "yellow", "purple", "green"],
            "height": [80, 80, 70, 100, 90, 80, 80, 70, 
                       100, 90, 91]}

new_df_2 = pd.DataFrame(new_data_2)

# joined outer: 

joined_outer_2 = first_10.join(new_df_2, how='outer') 

joined_outer_2
```

**Concatenating**  
You can also join two dataframes by simply gluing them together. 

```python
# We just made a subset of the original dataframe called 'first_10' 
# now we find the last 10 and glue them together

# selecting the last 10 rows using one of the methods we learned
last_10 = df.iloc[-10:, :]  

# Now we concatenate the two dataframes together 
# axis=0 means we concatenate rows, axis=1 would concatenate columns
concatenated = pd.concat( [first_10, last_10], axis=0) 
concatenated
```

**Music sales challenge**

Write a script that:

1. Combines the tables of best-selling physical singles and best-selling digital singles on the Wikipedia page "List_of_best-selling_singles"
2. Outputs the artist and single name for the year you were born. If there is no entry for that year, take the closest year after you were born.
3. Outputs the artist and single name for the year you were 15 years old.

```python
# Starter code
url_music = "https://en.wikipedia.org/wiki/List_of_best-selling_singles"

# Add a User-Agent header so Wikipedia doesn't block it
headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"}
response = requests.get(url_music, headers=headers)

# Pass the HTML text to pandas
musicdata = pd.read_html(StringIO(response.text))

#Extracting physical and digital singles from the musicdata
physical_singles = musicdata[0]
digital_singles = musicdata[3]

physical_singles["Type"] = "Physical"
digital_singles["Type"] = "Digital"

# Combining the two tables
combined_singles = pd.concat(
    [physical_singles, digital_singles],
    ignore_index=True
)

combined_singles.head()
```

```python
# Print the artist and single from the year you were born

#Hint: combined_singles[combined_singles['Year']] 
# finds the column with all the years songs were released:
singles_my_2004 = combined_singles[combined_singles
    ['Year'] == 2004.0]
singles_my_2004

# closest year to me: 
singles_my_2010 = combined_singles[combined_singles
    ['Year'] == 2010.0]

# print it out:
print(singles_my_2010[['Artist', 'Single', 'Year']])
```

```python
# Print the artist and single from the year you were 15 years old. 

# 15 year old:
singles_my_2019 = combined_singles[combined_singles
    ['Year'] == 2019.0]

# print it out:
print(singles_my_2019[['Artist', 'Single', 'Year']])
```

## Space challenge

1. Make a single dataframe that combines the space missions from the 1950's to the 2020's
2. Write a script that returns the year with the most launches
3. Write a script that returns the most common month for launches
4. Write a script that ranks the months from most launches to fewest launches

```python
# Starter code.
url_space =  "https://en.wikipedia.org/wiki/Timeline_of_Solar_System_exploration"

# Add a User-Agent header so Wikipedia doesn't block it
headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"}
response = requests.get(url_space, headers=headers)

# Pass the HTML text to pandas
spacedata = pd.read_html(StringIO(response.text))

# combine all tables into data frame
combined_space = pd.concat(spacedata, 
                           ignore_index = True)

# Dropping column we dont need
combined_space = combined_space.iloc[:, 0:3]
combined_space.head()
```

```python
## The year with the most launches 

# Hint:  we want the column 'Launch date', 
# only the rows with a date, and then converteed to a string

# select columns with launch dates: 
launch_dates = combined_space[["Launch date"]]

# remove NaN:
launch_dates = launch_dates.dropna()

# convert the dates: 
launch_dates["Launch date"] = pd.to_datetime(launch_dates
    ['Launch date'], format='%d %B %Y')

# get column with the year: 
launch_dates['Year'] = launch_dates['Launch date'].dt.year

# to get column with the month: 
launch_dates['Month'] = launch_dates['Launch date'].dt.month
launch_dates['Month'] = pd.to_datetime(launch_dates['Month'], 
                                       format='%m').dt.strftime('%b')

# to get the year with the most launches: 
# group launches by year to count them
launches_year = launch_dates.groupby('Year').size()
# find the max amount of launches by year
max_year = launches_year.max() 
# find the year that has the max amount of launches
years_with_max = launches_year[launches_year == max_year].index 

# use a for-loop to go through years (if more years with max launches): 
for year in years_with_max:
    print(f"The year {year} had the most launches at"
    f" {max_year} launches.") # print the message
```

```python
# The month with the most launches:

# count the max launches by month
most_launches_month = launch_dates['Month'].value_counts().max()
# finds the month
launches_month = launch_dates['Month'].value_counts().idxmax() 

# print the message
print(f"{launches_month} had the most launches at"
f" {most_launches_month} launches.") 
```

```python
# Ranking of months with the most to the fewest launches

most_launches_month = launch_dates['Month'
    ].value_counts() # counts launches by month

most_launches_month
```
